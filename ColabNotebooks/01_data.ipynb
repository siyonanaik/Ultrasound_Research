```python
import pandas as pd

manifest_path = (
    "/content/drive/MyDrive/Ultrasound_Project-Siyona/"
    "data/busi_manifest.csv"
)

manifest_df = pd.read_csv(manifest_path)

print("Manifest rows:", len(manifest_df))
print("\nClass counts:")
print(manifest_df["true_label"].value_counts())

manifest_df.head()
```

```python
from sklearn.model_selection import train_test_split

final_test_df, development_df = train_test_split(
    manifest_df,
    test_size=20,
    stratify=manifest_df["true_label"],
    random_state=42,
)

development_df = development_df.copy()
final_test_df = final_test_df.copy()

development_df["split"] = "development"
final_test_df["split"] = "test"

print("Development images:", len(development_df))
print(development_df["true_label"].value_counts())

print("\nUntouched test images:", len(final_test_df))
print(final_test_df["true_label"].value_counts())
```

```python
from pathlib import Path

splits_folder = Path(
    "/content/drive/MyDrive/Ultrasound_Project-Siyona/data/splits"
)
splits_folder.mkdir(parents=True, exist_ok=True)

development_path = splits_folder / "busi_development_20.csv"
test_path = splits_folder / "busi_test_760.csv"

development_df.to_csv(development_path, index=False)
final_test_df.to_csv(test_path, index=False)

print("Saved development split to:", development_path)
print("Saved test split to:", test_path)
```
